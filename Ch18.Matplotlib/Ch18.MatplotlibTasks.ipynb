**Task 1**:Complete the code below to create a image which looks exactly the same as the follow one:
![mat](img/task1.png)

Note: the first curve is a sine curve and the second curve is a cosine curve.

Note: the functions that may be used include: add_axes, figure, plot

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 50, 200)
y1 = np.sin(2 * np.pi * x / 25)
y2 = np.cos(2 * np.pi * x / 25)

plt.subplot(2, 1, 1)
plt.plot(x, y1)
plt.ylim(-2, 2)

plt.subplot(2, 1, 2)
plt.plot(x, y2)
plt.ylim(-2, 2)

plt.show()
```

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

lines = []
line_styles = ['-', '-.', '--', ':']
line_colors = ['red', 'blue', 'green', 'black']
x = np.linspace(0, 10, 1000)

phases = [0, -np.pi/2, -np.pi, -3*np.pi/2]
labels = ['part A', 'part B', 'part C', 'part D']

for phase, ls, lc, label in zip(phases, line_styles, line_colors, labels):
    y = np.sin(x + phase)
    line, = plt.plot(x, y, linestyle=ls, color=lc, label=label)
    lines.append(line)

leg1 = plt.legend(lines[:2], labels[:2], loc='upper right')
plt.gca().add_artist(leg1)

plt.legend(lines[2:], labels[2:], loc='lower right')


plt.show()
```

**Task 3**:Complete the code below to create a image which looks exactly the same as the follow one:
![mat](img/task3.png)

Note: the functions that may be used include:
plot, stackplot, xlabel, ylabel, title, legend

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]

sleeping = [7, 8, 6, 11, 7]
eating   = [2, 3, 4,  3, 2]
working  = [7, 8, 7,  2, 2]
playing  = [8, 5, 7,  8, 13]

# YOUR CODE HERE
labels = ['Sleeping', 'Eating', 'Working', 'Playing']
colors = ['purple', 'turquoise', 'red', 'black']

plt.stackplot(days,
              sleeping,
              eating,
              working,
              playing,
              labels=labels,
              colors=colors)

plt.legend(loc='upper right')
plt.xlabel('x')              
plt.ylabel('y')              
plt.title('Stack Plot')      

plt.show()
```

```python
import matplotlib.pyplot as plt
 

slices = [7,2,2,13]
activities = ['sleeping','eating','working','playing']
cols = ['c','m','r','b']
# YOUR CODE HERE
explode = (0, 0.1, 0, 0)
plt.pie(
    slices,
    labels=activities,
    colors=cols,
    explode=explode,
    autopct='%.1f%%',
    startangle=90,
    shadow=True
)
plt.title('Pie Plot')

plt.show()
```
