## The object-oriented API

The main idea with object-oriented programming is to 
* have objects that one can apply functions and actions on
* useful when more than one figure is created, or when a figure contains more than one subplot. 

Practice:
* create a figure instance in the `fig` variable
* create a new axes instance `axes` using the `add_axes` method:

```python
x = np.linspace(0, 5, 10)# what is linspace? What is x?
y = x ** 2 # What is y?

fig = plt.figure()

axes = fig.add_axes([0.1, 0.2, 0.8, 0.8])
# coordinate of the (left, bottom) point, and the size of the rectangle (width, height) (range 0 to 1)

axes.plot(x, y, 'r')

axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('hello world');
```

the advantage is that 
* we now have full control of where the plot axes are placed
* we can easily add more than one axes to the figure:

```python
fig = plt.figure()

axes1 = fig.add_axes([0.1, 0.2, 0.8, 0.8]) # main axes
axes2 = fig.add_axes([0.2, 0.5, 0.4, 0.3]) # insert axes

# main figure
axes1.plot(x, y, 'r')
axes1.set_xlabel('x')
axes1.set_ylabel('y')
axes1.set_title('title')

# insert figure
axes2.plot(y, x, 'g')
axes2.set_xlabel('y')
axes2.set_ylabel('x')
axes2.set_title('insert title');
```

### Formatting text: $LaTeX$, fontsize

* Matplotlib has great support for $LaTeX$. 

* Use dollar signs `$` encapsulate $LaTeX$ in any text (legend, title, label, etc.). 
* For example, `r"$y=x^3$"`.


```python
fig, ax = plt.subplots()

ax.plot(x, x**2, label=r"$y = \alpha^2$")
ax.plot(x, x**3, label=r"$y = \alpha^3$")
ax.legend(loc=2, fontsize=20) # upper left corner
ax.set_xlabel(r'$\alpha$', fontsize=20)
ax.set_ylabel(r'$y$', fontsize=20)
ax.set_title('title', fontsize=20);
```

## 3D figures

* Create an 3D axes by passing a `"projection": "3d"` keyword argument to the `add_axes` or `subplots` methods.
* Make Z values besides X and Y values
* `plot_surface` is used to create surface figure
* `plot_wireframe` is used to create wire-frame figure

#### Surface plots

```python
from matplotlib import cm
from matplotlib.ticker import LinearLocator

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

# Make data.
X = np.arange(-5, 5, 0.25) # what is X?
Y = np.arange(-5, 5, 0.25)

X, Y = np.meshgrid(X, Y) #what is meshgrid?
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

# Plot the surface.
surf = ax.plot_surface(X, Y, Z, cmap=cm.coolwarm,
                       linewidth=0, antialiased=False)

# Add a color bar which maps values to colors.
fig.colorbar(surf, shrink=0.5, aspect=5)
```

#### Wire-frame plot

```python
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(1, 1, 1, projection='3d')
p = ax.plot_wireframe(X, Y, Z, rstride=1, cstride=1)
```

## Animations

* Matplotlib includes a simple API for generating animations for sequences of figures. 
* The `FuncAnimation` method generate a movie from sequences of figures. 
Its arguments include
** `fig`, a figure canvas, 
** `func`, a function that we provide updates of the figure, 
** `init_func`, a function we provide to initialize the figure, 
** `frames`, data to pass `func` for each frame of the animation 
** `blit`, tells the animation function to only update parts of the frame which have changed (for smoother animations).


To use the animation features in matplotlib we first need to import the module `matplotlib.animation`:

#### Example: animation for sine curve

```python
# Enable interactive plot
%matplotlib notebook

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

fig, ax = plt.subplots()
xdata, ydata = [], []
ln, = ax.plot([], [], 'ro')

def init():
    ax.set_xlim(0, 2*np.pi) #Set the x-axis view limits.
    ax.set_ylim(-1, 1) #why set the y data between -1 and 1?.
    return ln,

def update(frame):
    xdata.append(frame)
    ydata.append(np.sin(frame))
    ln.set_data(xdata, ydata)
    return ln,

ani = FuncAnimation(fig, update, 
                    frames=np.linspace(0, 2*np.pi, 128),
                    init_func=init, blit=True)
plt.show()
```

### Backends

Matplotlib has a number of "backends" which are responsible for rendering graphs. The different backends are able to generate graphics with different formats and display/event loops. There is a distinction between noninteractive backends (such as 'agg', 'svg', 'pdf', etc.) that are only used to generate image files (e.g. with the `savefig` function), and interactive backends (such as Qt4Agg, GTK, MaxOSX) that can display a GUI window for interactively exploring figures. 

A list of available backends are:

```python
print(matplotlib.rcsetup.all_backends)
```

The default backend, called `agg`, is based on a library for raster graphics which is great for generating raster formats like PNG.

Normally we don't need to bother with changing the default backend; but sometimes it can be useful to switch to, for example, PDF or GTKCairo (if you are using Linux) to produce high-quality vector graphics instead of raster based graphics.

#### The IPython notebook inline backend

In new notebooks created for IPython 1.0+, I would recommend using `%matplotlib inline`, since it is tidier and you have more control over which packages are imported and how. ### Colormap and contour figures

Colormaps and contour figures are useful for plotting functions of two variables. In most of these functions we will use a colormap to encode one dimension of the data. There are a number of predefined colormaps. It is relatively straightforward to define custom colormaps. For a list of pre-defined colormaps, see: http://www.scipy.org/Cookbook/Matplotlib/Show_colormaps

```python
alpha = 0.7
phi_ext = 2 * np.pi * 0.5

def flux_qubit_potential(phi_m, phi_p):
    return 2 + alpha - 2 * np.cos(phi_p) * np.cos(phi_m) - alpha * np.cos(phi_ext - 2*phi_p)
```

```python
phi_m = np.linspace(0, 2*np.pi, 100)
phi_p = np.linspace(0, 8*np.pi, 100)
X,Y = np.meshgrid(phi_p, phi_m)
Z = flux_qubit_potential(X, Y).T
```

#### imshow

```python
fig, ax = plt.subplots()

# vmin: minimum value; extent: extent of x, y coordinates.
im = ax.imshow(Z, cmap=matplotlib.cm.RdBu, vmin=abs(Z).min(), vmax=abs(Z).max(), extent=[0, 1, 0, 1])
im.set_interpolation('bilinear')

cb = fig.colorbar(im, ax=ax)
```

#### contour

```python
fig, ax = plt.subplots()

cnt = ax.contour(Z, cmap=matplotlib.cm.RdBu, vmin=abs(Z).min(), vmax=abs(Z).max(), extent=[0, 1, 0, 1])
```
