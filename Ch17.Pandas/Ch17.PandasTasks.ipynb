**Task 1**: In the code cell below, complete the function `aboveAverage` which takes a series `s` as an arguement, and returns a series which contains all values larger than the average in its `s` original order.

```python
import pandas as pd

def aboveAverage(s):
    avg = s.mean()
    return s[s > avg].reset_index(drop=True)
```

```python
s = pd.Series([1,5,3,6,8])  
t = pd.Series([5,6,8])
assert(t.equals(aboveAverage(s)))

s = pd.Series([11,12,13,14])  
t = pd.Series([13,14])
assert(t.equals(aboveAverage(s)))
```

**Task 2**:  In the code cell below, complete the function `twoSum` which takes a series `s` and an integer `target` as arguements, and returns a list which contains the indices of the two numbers such that they add up to `target`.

Example:

Input: s = [2,7,11,15], target = 9

Output: [0,1]

Output: Because nums[0] + nums[1] == 9, we return [0, 1].

```python
import pandas as pd
def twoSum(s, target):
    seen = {}
    for i, num in enumerate(s):
        # Calculate the complement of the current number
        complement = target - num
        # Check if the complement already exists in the dictionary
        if complement in seen:
            return [seen[complement], i]
        # Store the index of the current number
        seen[num] = i
```

```python
s = pd.Series([1,5,3,6,8])  
t = twoSum(s,7)
assert(s[t[0]] + s[t[1]] == 7)

s = pd.Series([3,2,4])  
t = twoSum(s,6)
assert(s[t[0]] + s[t[1]] == 6)
```

**Task 3**: In the code cell below, complete the function `tranposeDataFrame` which takes a dataFrame df as an arguement, and returns a two-dimensional list which is the transpose of the dataFrame. The label and index are ignored in the output.

```python
import pandas as pd
def tranposeDataFrame(df):
    return df.values.T.tolist()
```

```python
df1 = pd.DataFrame({'c1': [1, 4,7],
                   'c2': [2, 5,8],
                   'c3': [3, 6,9]})
t = tranposeDataFrame (df1)

assert (t == [[1, 4, 7], [2, 5 ,8], [3, 6 ,9]])

df1 = pd.DataFrame({'c1': [1, 2,3],
                   'c2': [4, 5,6]})
t = tranposeDataFrame (df1)

assert (t == [[1, 2, 3], [4, 5, 6]])
```

**Task 4**: In the code cell below, complete the function `setDataFrameZeros` which takes a dataFrame `df` as an arguement, and returns a new dataFrame. The label and index in the new dataFrame is the same as those in `df`.  If an element is 0 in  `df`, set its entire row and column to 0 in the new dataFrame. All other elements are the same as those in  `df`.

Example 1:


![pd spiral1](img/mat1.jpg)



Example 2:

![pd spiral1](img/mat2.jpg)

```python
import pandas as pd
def setDataFrameZeros(df):
    new_df = df.copy()
    zero_rows = (df == 0).any(axis=1)
    zero_cols = (df == 0).any(axis=0)
    new_df.loc[zero_rows, :] = 0
    new_df.loc[:, zero_cols] = 0
    return new_df
```

```python
df1 = pd.DataFrame({'c1': [1, 4,7],
                   'c2': [2, 0,8],
                   'c3': [3, 6,9]})

df2 = pd.DataFrame({'c1': [1, 0,7],
                   'c2': [0, 0,0],
                   'c3': [3, 0,9]})

assert (df2.equals(setDataFrameZeros(df1)))

df1 = pd.DataFrame({'c1': [0, 3,1],
                   'c2': [1, 4,3],
                   'c3': [2, 5,1],
                   'c4': [0, 2,5]})

df2 = pd.DataFrame({'c1': [0, 0,0],
                   'c2': [0, 4,3],
                   'c3': [0, 5,1],
                   'c4': [0, 0,0]})

assert (df2.equals(setDataFrameZeros(df1)))
```

**Task 5**: In the code cell below, complete the function `calculateFinalScore` which takes a dataFrame `df` with columns for student names, test scores, and homework scores, and calculates the final score for each student based on a weighted average of their test scores and homework scores, i.e., final_score = 0.6 * test_scores + 0.4 * homework_scores. Your returned dataframe should contain a new column called `final_scores`.

```python
import pandas as pd

def calculateFinalScore(df):
    result = df.copy()
    result['final_scores'] = 0.6 * result['test_scores'] + 0.4 * result['homework_scores']
    return result
```

```python
df = pd.DataFrame({
    'student_names': ['Alice', 'Bob', 'Charlie'],
    'test_scores': [80, 90, 70],
    'homework_scores': [90, 80, 100]
})

expected = pd.DataFrame({
    'student_names': ['Alice', 'Bob', 'Charlie'],
    'test_scores': [80, 90, 70],
    'homework_scores': [90, 80, 100],
    'final_scores': [84.0, 86.0, 82.0]
})

assert calculateFinalScore(df).equals(expected)
```

```python
df = pd.DataFrame({
    'student_names': ['David', 'Emily', 'Frank'],
    'test_scores': [70, 80, 90],
    'homework_scores': [80, 90, 70]
})
expected = pd.DataFrame({
    'student_names': ['David', 'Emily', 'Frank'],
    'test_scores': [70, 80, 90],
    'homework_scores': [80, 90, 70],
    'final_scores': [74.0, 84.0, 82.0]
})
assert calculateFinalScore(df).equals(expected)
```
