### Lines and Indentation

In Python, end of a statement is marked by a newline character (no semicolon!). But we can write **a single statement over multiple lines** with the line continuation character (\\). For example:

```python
# use "\" to divide one statement over multiple lines
a = 1 + 2 + 3 + \
    4 + 5 + 6 + \
    7 + 8 + 9
```

Further, line continuation is implied inside parentheses (()), brackets ([]) and braces ({}). In the following example, the line continuation character can be omitted.

```python
print("Hello",
      "World")
gender = ["female",
         "male"]
```
         

If you want to put multiple statements in a single line, seperate them using semicolons (;).

```python
a = 1; b = 2; c = 3
```

Python does not use braces ({}) to indicate blocks of code. Code blocks are denoted by line indentation and this is rigidly enforced. ***Wrong indentation will cause error!***

A code block starts with indentation and ends with the first unindented line. 
Generally *four whitespaces* are used for indentation. In a Jupyter notebook, you may press "tab" to input four whitespaces.

````python
# Example of correct indentation

def printRange(n):
    for i in range(1, n):
        print(i)
        if i % 5 == 0:
            print("--------------")
        if i == 99:
            print("99 is enough.")
            break

myNumber = 12
print("Printing numbers:")
printRange(myNumber)
````

**Task 9**: Complete the function `triangle` which accepts a positive integer, `n`, as argument, and returns a string representing a triangle of stars containing `n` lines. For example, if `n=5`,  the triangle should be like this:
```
        * 
       * * 
      * * * 
     * * * * 
    * * * * * 
```

*Note: Every line of the triangle should end with a new line character.*

````python
def triangle(n): 
    res = ""
    
    for i in range(1, n + 1):
        res += " " * (n - i)
        res += "* " * i
        res = res.rstrip() + "\n"
    
    return res
    
# Driver Code 
n = 5
print(triangle(n))
````

````python
# This cell contains test cases. Do not modify.
assert triangle(1) == '*\n'
assert triangle(2) == ' *\n* *\n'
assert triangle(5) == '    *\n   * *\n  * * *\n * * * *\n* * * * *\n'
````

**Task 10**: Complete the function `isPrime` which accepts a positive integer, `x`, as argument, and decides whether `x` is a prime number. If the answer is yes, the function should return `True`, and `False` otherwise. 

*A prime number is a natural number greater than 1 that is not a product of two smaller natural numbers.*

````python
def isPrime(x):
    if (x == 1): return False;
    if (x == 2): return True;

    d = 2;
    while (d * d <= x):
        if (x % d == 0): return False;
        
        d += 1;

    return True;
````

````python
# This cell contains test cases. Do not modify.
assert isPrime(1) == False
assert isPrime(5) == True
assert isPrime(6) == False
````

**Task 11**: In the code cell below, complete the function `triangleArea` which returns the area of a triangle. The three arguments of the function, `a` ,`b`, and `c`, are the length of the edges. `area` should return 0 if the arguments are invalid.

*Hint 1: Use Heron’s Formula.*

*Hint: You may use `x**y` to compute the $x^y$, and `y` can be any float number. Consider using it to compute the square root of a number.* 

````python
def triangleArea(a, b, c):
    if a <= 0 or b <= 0 or c <= 0 or (a + b <= c) or (a + c <= b) or (b + c <= a):
        return 0
    
    s = (a + b + c) / 2
    area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
    
    return area
````

````python
# This cell contains test cases. Do not modify.
assert triangleArea (-3, -3, -3) == 0
assert triangleArea (3, 3, 7) == 0
assert triangleArea (3, 4, 5) ==  6
````
