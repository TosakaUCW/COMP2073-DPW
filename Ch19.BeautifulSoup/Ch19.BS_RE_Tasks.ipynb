**Task 1:** In the code cell below, complete function `tableRowCounter` take the parameter `s` which is a string of HTML as input and returns the number of rows in the first table of `s`. The row that contains the table header should not be counted.

```python
import bs4
from bs4 import BeautifulSoup

def tableRowCounter(s):
    
    soup = BeautifulSoup(s, 'html.parser')
    table = soup.find('table')
    if table is None:
        return 0
    rows = table.find_all('tr')
    count = 0
    for tr in rows:
        if tr.find('th'):
            continue
        count += 1
    return count
```

```python
s ="<body><table><tr><th>c1</th><th>c2</th></tr><tr><td>1</td><td>1</td></tr></table><body>"

assert tableRowCounter(s)==1

s ="<body><table><tr><td>1</td><td>1</td></tr><tr><td>1</td><td>1</td></tr><tr><td>1</td><td>1</td></tr></table><body>"
assert tableRowCounter(s)==3

s ="<body><table><tr><td>1</td><td>1</td></tr><tr><th>c1</th><th>c2</th></tr><tr><td>1</td><td>1</td></tr></table><body>"
assert tableRowCounter(s)==2
```

**Task 2:** In the code cell below, complete function `list2table` take the parameter `s` which is a string of HTML as input and return a string. `list2table` extracts the data from all `ordered lists` in `s` and converts each of them to a table. The data table in each item is seperated by comma. For example:
![list2table](Files/list2table.png)

```python
import re

def lists2tables(s):
    def repl(m):
        inner = m.group(1)
        items = re.findall(r'<li>(.*?)</li>', inner)
        rows = [[cell.strip() for cell in item.split(',')] for item in items]
        if not rows:
            return ''
        max_cols = max(len(r) for r in rows)
        parts = ['<table border="1">']
        for idx, row in enumerate(rows, start=1):
            parts.append('<tr>')
            parts.append(f'<td>{idx}</td>')
            for cell in row:
                parts.append(f'<td>{cell}</td>')
            for _ in range(max_cols - len(row)):
                parts.append('<td></td>')
            parts.append('</tr>')
        parts.append('</table>')
        return ''.join(parts)
    return re.sub(r'<ol>(.*?)</ol>', repl, s, flags=re.DOTALL)
```

```python
s= "<body><ol><li>coffee,milk,water</li><li>banana,apple</li><li>tea</li></ol></body>"
s1 = "<body><table border=\"1\"><tr><td>1</td><td>coffee</td><td>milk</td><td>water</td></tr><tr><td>2</td><td>banana</td><td>apple</td><td></td></tr><tr><td>3</td><td>tea</td><td></td><td></td></tr></table></body>"
assert lists2tables(s) == s1
s= "<body><ol><li>coffee,milk,water</li><li>banana,apple</li><li>tea</li></ol><ol><li>banana,apple</li><li>tea</li></ol></body>"
s1 = "<body><table border=\"1\"><tr><td>1</td><td>coffee</td><td>milk</td><td>water</td></tr><tr><td>2</td><td>banana</td><td>apple</td><td></td></tr><tr><td>3</td><td>tea</td><td></td><td></td></tr></table><table border=\"1\"><tr><td>1</td><td>banana</td><td>apple</td></tr><tr><td>2</td><td>tea</td><td></td></tr></table></body>"
assert lists2tables(s) == s1
```

**Task 4 [Regular Expressions]:** In the code cell below, complete function `validIPAdd` take the parameter `s` which is a string of IP address and returns true if it is a valide IP address and false otherwise. A valid IP address contains four numbers separated by `.`, i.e., `a.b.c.d`, where `a`, `b`, `c` and `d` are integers between 0 and 255.

```python
import re
def validIPAdd(s):
    pattern = re.compile(
        r'^(?:25[0-5]|2[0-4]\d|[01]?\d?\d)'
        r'(?:\.(?:25[0-5]|2[0-4]\d|[01]?\d?\d)){3}$'
    )
    return bool(pattern.match(s))
```

```python
assert validIPAdd('125.122.122') == False
assert validIPAdd('125.122.122.12') == True
assert validIPAdd('256.122.122.12') == False
assert validIPAdd('256.122.122.12') == False
```

**Task 5 [Regular Expressions]:** In the code cell below, complete function `strongPassWord` take the parameter `s` which is a string of password and returns true if it is a strong password and false otherwise. A password is strong if it contains 8-10 letters, It must include digits, lowercase and uppercase characters and nothing else.

```python
import re
def strongPassWord(s):
    pattern = re.compile(
        r'^(?=.*[0-9])(?=.*[a-z])(?=.*[A-Z])[A-Za-z0-9]{8,10}$'
    )
    return bool(pattern.match(s))
```

```python
assert strongPassWord('abc') == False
assert strongPassWord('ab123ABCD') == True
assert strongPassWord('ABC124~~~') == False
assert strongPassWord('ABCABC123') == False
```
