In Python, a string is a sequence of Unicode characters. Unicode was introduced to include every character in all languages and bring uniformity in encoding. You can learn about Unicode from [Python Unicode](http://docs.python.org/3.3/howto/unicode.html).

Note that Python handles a single character as a string consisting one character. There is not a type in python equivalent to the data type `char` in C.

## 1. Create a string

Strings can be created by enclosing characters inside a pair of single quotes or double-quotes. Even triple quotes can be used in Python but generally used to represent multiline strings and docstrings. Below are some example of string creation.

```python
# defining strings in Python
# all of the following are equivalent
my_string = 'Hello'
print(my_string)

my_string = "Hello"
print(my_string)

my_string = '''Hello'''
print(my_string)

# triple quotes string can extend multiple lines
my_string = """Hello,
world!"""
print(my_string)

# escape sequences can be used to indicate a new line character, too
my_string = 'Hello, \nworld!'
print(my_string)
```

### 1.1 Raw Strings

If you need to specify some strings where no special processing such as escape sequences are handled, then what you need is to specify a raw string by prefixing r or R to the string.

```python
raw = r'Hello, \nworld!'
print(raw)

raw = R'Hello, \nworld!'
print(raw)
```

### 1.2 Unicode Strings

Unicode is a standard way of writing international text. If you want to write text in your native language such as Chinese, then you need to have a Unicode-enabled text editor. Similarly, Python allows you to handle Unicode text - all you need to do is to prefix u or U to the string.

```python
s = u'为革命保护视力'
print(s)
```

## 2. Access a string

We can access individual characters using *indexing* and a range of characters using *slicing*. For a string of $n$ characters, the index starts from $0$ and ends at $n−1$.

Further, Python allows negative indexing for its sequences. The index of -1 refers to the last item, -2 to the second last item and so on. Therefore, for a string of $n$ characters, the negative index starts from $-1$ and ends at $-n$. Below shows the index and the negative index of a string "python3":

|String: |p|y|t|h|o|n|3|
|---|---|---|---|---|---|---|---|
|Index:|0|1|2|3|4|5|6|
|Negative Index:|-7|-6|-5|-4|-3|-2|-1|


Trying to access a character out of the index range will raise an IndexError. Last, the index must be an integer. We can't use floats or other types, this will result into TypeError.

```python
#Accessing string characters in Python
my_str = 'python3'
print('my_str = ', my_str)

#first character
print('my_str[0] = ', my_str[0])

#last character
print('my_str[-1] = ', my_str[-1])

#slicing my_str[0] to my_str[4]
print('my_str[:5] = ', my_str[:5])

#slicing the last two characters of string
print('my_str[-2:] = ', my_str[-2:])

#slicing my_str[1] to my_str[4]
print('my_str[1:5] = ', my_str[1:5])

#slicing my_str[5] to my_str[-3], the result is empty
print('my_str[5:-2] = ', my_str[5:-2])
```

## 3. String Operations

There are many operations that can be performed with strings which makes it one of the most used data types in Python.

### 3.1 Concatenation of Two or More Strings

Joining of two or more strings into a single one is called concatenation.

The `+` operator does this in Python. Simply writing two string literals together also concatenates them.

The `*` operator can be used to repeat the string for a given number of times.


```python
# Python String Operations
str1 = 'Hello'
str2 ='World!'

# Concatenation using +
print('str1 + str2 = ', str1 + str2)

# Concatenation by writing two string literals together
str3 = "Hello""World!"
print(str3)

# Repeat a string using *
print('str1 * 3 =', str1 * 3)
```

### 3.2 Iterating Through a string
We can iterate through a string using a `for` loop. Here is an example to count the number of `'l'`s in a string.

```python
# Iterating through a string
count = 0
for letter in 'Hello World':
    if(letter == 'l'):
        count += 1
print(count,'letters found')
```

### 3.3 String Membership Test
We can test if a substring exists within a string or not, using the keyword in.

```python
my_str = 'python'
print('py' in my_str)
print('pn' not in my_str)
```

### 4.1 Splits

`split()` splits a string into a list of sub-strings. The default delimiter for the split is `space`, but it can be specified.

```python
# the default delimiter for split is space
s = 'Once a rocker, always a rocker.'
splits = s.split()
print (splits)

# the delimiter is ','
s = 'Once a rocker, always a rocker.'
splits = s.split(',')
print (splits)
```

### 4.2 Strip

The `strip()` method returns a copy of the string by removing both the leading and the trailing characters (based on the string argument passed).

```python
s = '\n   .. a baby is a baby.  \n'

# By default, the whitespaces and new line characters are stripped
s = s.strip()
print (s)

# '.' is stripped
s = s.strip('.')
print (s)

# any of the four characters in 'a', 'b', 'y', ' ' is tripped
s = s.strip('aby ')
print (s)
```

### 4.3 Find

The `find()` method returns the index of the first occurrence of the substring (if found). If not found, it returns -1.

```python
# the inde of the first occurrrence is returned
s = 'Once a rocker, always a rocker.'
print(s.find('rocker'))

# find() returns -1 if no such substring exists
print(s.find('Rocker'))
```

**Task 1:** In the code cell below, complete the function `words` which accepts a `sentence` and returns a list of words that appear in the senence. The words are separated with spaces. Further, the sentence may contain punctuations including `,`, `.`, and `"`, and the words in your list should not.

For example, the result for `words('Farewell, "team leader".')` should be `['Farewell', 'team', 'leader']`

```python
# Code for Task 1
def words(sentence):
    res = []
    now = ""
    for ch in sentence:
        if ('a' <= ch and ch <= 'z' or 'A' <= ch and ch <= 'Z'):
            now += ch;
        else:
            if (now != ""):
                res.append(now);
            now = "";
    if (now != ""):
        res.append(now);
    return res;
```

```python
assert words('Farewell, "team leader".') == ['Farewell', 'team', 'leader']
assert words('There, there.') == ['There', 'there']
assert words('Happy cat') == ['Happy', 'cat']
assert words('') == []
```

**Task 2:** In the code cell below, complete the function `standardTime` which accepts a time string, `time24`, in the 24-hour format and returns the same time string in the 12-hour format.

```python
# Code for Task 2
def standardTime(time24):
    """
    Precondition: time24, a time in the 24-hour (military) format.
    Returns: the time in the 12-hour (standard) format.
    
    the 24-hour format has the form '<hours>:<minutes>'. The 
    hours are between 0 and 23, and are always two digits
    (so there may be a leading zero).  The minutes are 
    between 0 and 59, and are always 2 digits.
    
    the 12-hour format has the form '<hours>:<minutes> AM' or
    '<hours>:<minutes> PM'.  There is a space before the AM
    or PM.  The hours must be between 1 and 12, and there
    is NO leading zero (so it may either be 1 or 2 digits).
    The minutes are between 0 and 59, and are always 2 digits.
    
    Examples:
        '14:45' becomes '2:45 PM'
        '09:05' becomes '9:05 AM'
        '00:00' becomes '12:00 AM'
        '12:10' becomes '12:10 PM'
    """
    hours, minutes = map(int, time24.split(':'))
    
    if hours < 12:
        period = 'AM'
        if hours == 0:
            hours = 12
    else:
        period = 'PM'
        if hours > 12:
            hours -= 12
    
    time12 = f"{hours}:{minutes:02d} {period}"
    
    return time12
```

```python
assert standardTime('14:45') == '2:45 PM'
assert standardTime('09:05') == '9:05 AM'
assert standardTime('00:00') == '12:00 AM'
assert standardTime('12:10') == '12:10 PM'
```

**Task 3:** In the code cell below, complete the function `militaryTime` which accepts a time string, `time12`, in the 12-hour format and returns the same time string in the 24-hour format. The definition of the two time formats is the same as described in **Task 2**.

```python
# Code for Task 3
def militaryTime(time12):
    """
    Precondition: time12, a time in the 12-hour (standard) format.
    Returns:  a time in the 24-hour (military) format.
    
    the 24-hour format has the form '<hours>:<minutes>'. The 
    hours are between 0 and 23, and are always two digits
    (so there may be a leading zero).  The minutes are 
    between 0 and 59, and are always 2 digits.
    
    the 12-hour format has the form '<hours>:<minutes> AM' or
    '<hours>:<minutes> PM'.  There is a space before the AM
    or PM.  The hours must be between 1 and 12, and there
    is NO leading zero (so it may either be 1 or 2 digits).
    The minutes are between 0 and 59, and are always 2 digits.
    
    Examples:
        '2:45 PM' becomes '14:45'
        '9:05 AM' becomes '09:05'
        '12:00 AM' becomes '00:00'
        '12:10 PM' becomes '12:10'
    """
    time_part, period = time12.split()
    
    hours, minutes = map(int, time_part.split(':'))
    
    if period == 'PM' and hours != 12:
        hours += 12
    elif period == 'AM' and hours == 12:
        hours = 0
    
    time24 = f"{hours:02d}:{minutes:02d}"
    
    return time24
```

```python
assert militaryTime('2:45 PM') == '14:45'
assert militaryTime('9:05 AM') == '09:05'
assert militaryTime('12:00 AM') == '00:00'
assert militaryTime('12:10 PM') == '12:10'
```

**Task 4:** In the code cell below, complete the function `checkPasswd` which checks whether the string `passwd` is a valid password or not. A password is valid if

* Rule 1: It contains no less than 8 characters
* Rule 2: It contains at least 1 digit, 1 lower-case letter and 1 upper-case letter
* Rule 3: It contains only digits and english letters

The function returns `True` if the `passwd` is valid and `False` otherwise.

For example, 
* `checkPasswd('aA123456')` returns `True`
* `checkPasswd('aA12345')` returns `False` because it violates Rule 1.
* `checkPasswd('aa123456')` returns `False` because it violates Rule 2.
* `checkPasswd('aA123456***')` returns `False` because it violates Rule 3.

```python
# Code for Task 4
def checkPasswd(passwd):
    if len(passwd) < 8:
        return False
    has_digit = any(char.isdigit() for char in passwd)
    has_lower = any(char.islower() for char in passwd)
    has_upper = any(char.isupper() for char in passwd)
    if not (has_digit and has_lower and has_upper):
        return False
    if not all(char.isalnum() for char in passwd):
        return False
    return True
```

```python
assert checkPasswd('aA123456') == True
assert checkPasswd('aA12345') == False
assert checkPasswd('aa123456') == False
assert checkPasswd('aA123456***')== False
```

**Task 5:**  In the code cell below, complete the function `isPalindrome` which checks whether a `word` is a palindrome or not.

A palindrome is a word that reads the same backward as forward, regardless of the case. For example, `'Anna'` and `'racecar'` are palindromes.

If the word is empty, `isPalindrome` should return `True`.

*Challenge: Can you complete the function with one line of code?*

```python
# Code for Task 5
def isPalindrome(word):
    s = word.lower()
    return s == s[::-1];
```

```python
assert isPalindrome('Anna') == True
assert isPalindrome('racecar') == True
assert isPalindrome('') == True
assert isPalindrome('baby') == False
assert isPalindrome('ooxx') == False
```
