**Task 4**: Given an integer array `nums`, find the contiguous subarray within an array (containing at least one number) which has the largest product. You may have to realize the impact of negative values.

Example 1:

Input: [2,3,-2,4]

Output: 6

Explanation: [2,3] has the largest product 6.

Example 2:

Input:[-2,0,-1]

Output: 0

Explanation: The result cannot be 2, because [-2,-1]is not a subarray.

```python
import numpy as np
def maxProduct(nums):
    max_prod = nums[0]
    min_prod = nums[0]
    result = nums[0]
    
    for x in nums[1:]:
        if x < 0:
            max_prod, min_prod = min_prod, max_prod
        
        max_prod = max(x, max_prod * x)
        min_prod = min(x, min_prod * x)
        
        result = max(result, max_prod)
    
    return result
```

```python
assert maxProduct(np.array([2,3,-2,4])) == 6
assert maxProduct(np.array([-2,0,-1])) == 0
```

**Task 5**: You are given an n x n 2D matrix representing an image, rotate the image by 90 degrees (clockwise).

You have to rotate the image in-place, which means you have to modify the input 2D matrix directly. DO NOT allocate another 2D matrix and do the rotation.

Example 1:
<img src="img/mat1.jpg" width="400px">


* Input: matrix = [[1,2,3],[4,5,6],[7,8,9]]
* Output: [[7,4,1],[8,5,2],[9,6,3]]

Example 2:
<img src="img/mat2.jpg" width="400px">

* Input: matrix = [[5,1,9,11],[2,4,8,10],[13,3,6,7],[15,14,12,16]]

* Output: [[15,13,2,5],[14,3,4,1],[12,6,8,9],[16,7,10,11]]

Constraints:

1. n == matrix.length == matrix[i].length
2. 1 <= n <= 20
3. -1000 <= matrix[i][j] <= 1000

```python
import numpy as np
def rotate(A):
    n = A.shape[0]
    for i in range(n):
        for j in range(i + 1, n):
            A[i, j], A[j, i] = A[j, i], A[i, j]
    for i in range(n):
        A[i, :] = A[i, ::-1]
    return A
```

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
matrix = np.array(matrix)
assert np.array_equal(rotate(matrix) , np.array([[7, 4, 1],[8, 5, 2], [9, 6, 3]]))

matrix = [[5,1,9,11],[2,4,8,10],[13,3,6,7],[15,14,12,16]]
matrix = np.array(matrix)
assert np.array_equal(rotate(matrix),np.array([[15, 13,  2,  5],[14,  3,  4,  1],[12,  6,  8,  9],[16,  7, 10, 11]]))

matrix = [[5]]
matrix = np.array(matrix)
assert np.array_equal(rotate(matrix),np.array([[5]]))
```
