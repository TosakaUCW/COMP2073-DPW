## 2. Class Attributes and Instance Attributes

* An attribute, feature, quality, or characteristic of something or someone. Usually an "attribute" is used to denote a specific ability or characteristic which something or someone has.
* **Instance attributes** are owned by the specific instances of a class. This means for two different instances the instance attributes are usually different. They are similar to the <i>"instance variables"</i> in Java.
* **Class attributes** are attributes which are owned by the class itself. They will be shared by all the instances of the class. Therefore they have the same value for every instance. They are similar to the <i>"static variables"</i> in Java. We define class attributes outside of all the methods, usually they are placed at the top, right below the class header.

Below is an example of **Class Attributes**.

#### Example 2.1

```python
class A:
    # a class attribute is defined outside of all the methods
    a = "I am a class attribute!"
    
# class attributes can be accessed through the class name
print(A.a)
# class attributes can be accessed through objects
x = A()
print(x.a)
y = A()
print(y.a)
```

A class attribute belongs to the class. It is shared by all the objects of the class.

### 3.2 `__del__` method

The **\__del\__** method is called "destructor", and is evoked as soon as when the instance is about to be destroyed.

#### Example 3.2

```python
class Employee:
    
    empCount = 0 # class variable

    def __init__(self, name, salary): # Initialize the object
        self.name = name # object variable
        self.salary = salary # object variable
        Employee.empCount += 1
   
    def displayCount(): 
        print ("Total Employee %d" % Employee.empCount)
        
    def sayHi(self):
        print (self.name, 'says hello world')

    def displaySelf(self):
        print ("%s's salary is: %d"% (self.name, self.salary))
    
    def __del__(self):
        print ('%s is no longer with us'% self.name)
        Employee.empCount -= 1 
```

```python
Bob = Employee('Bob', 100)
```

## 4. Garbage Collection

* Python deletes unneeded objects (built-in types or class instances) automatically to free the memory space. The process by which Python periodically reclaims blocks of memory that no longer are in use is termed **Garbage Collection**.

* Python's garbage collector runs during program execution and is triggered when an object's **reference count** **reaches zero**.

* An object's reference count increases when it is 
    * assigned a new name or 
    * placed in a container (list, tuple, or dictionary).
    
*  The object's reference count decreases when it is
    * deleted with *del*
    * its reference is reassigned
    * its reference goes out of scope
    
* Garbage collection is extremely important in systems which run continuously over long term, where lots of unused variables might be accumulated which harm the system performance.

#### Example 4.1

```python
a = 1
b = a
a = 2
print (a)
print (b)
```

#### Example 4.2

```python
Tom = Employee('Tom', 50) # Create object "Tom"
Tom_copy = Tom # Add a new reference to "Tom"
Tom_group = [Tom] # Add a new reference to "Tom" by putting him to list

Tom = 1 # decrease Tom reference count by 1
Tom_copy = Employee('Jimmy', 200) # decrease Tom reference by reassigning to "Jimmy"
# decrease Tom reference by 1. Now "Tom" has no reference, then the destructor is called.
Tom_group[0] = 2 

```

* You normally will not notice when the garbage collector destroys an orphaned instance and reclaims its space.

## 5. Inheritance

### 5.1 Inheritance Basic

* One of the major benefits of object oriented programming is **reuse** of code and one of the ways this is achieved is through the **inheritance** mechanism.

* Suppose you want to write a program which has to keep track of the teachers and students in a college. They have some **common** characteristics such as name, age and address. They also have **specific** characteristics such as (salary, courses and leaves) for teachers and, (marks and fees) for students.

* You can create two independent classes "teacher" and "student" for each type and process them seperately. However, if you are required to implement a common function which both of two classes need, it is not good to implement the same function twice in two classes. The side effect is, every time you want to update information w.r.t this function, you need to update twice in the two classes, which is redundant and may cause inconsistency.

* A better way would be to create a common class called SchoolMember and then have the teacher and student classes **inherit** from this class. The "SchoolMember" is also called as *"base class"/"super class"/"parent class"*. The classes "student" and "teacher" are often called *"derived classes"/"subclasses"/"child classes"*.

#### Example 5.1

```python
class SchoolMember: # base class
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def follow_UIC_rules(self):
        print ('Every one in UIC should follow UIC rules 2018 version (updated)')      


class Faculty(SchoolMember): # subclass, inherit from "SchoolMember" class
    def facultyFunc(self):
        print ('I am a faculty function')


class Student(SchoolMember):
    def studentFunc(self):
        print ('I am a student function')
```

```python
f = Faculty('Eric', 30)
s = Student('Bob', 20)
f.follow_UIC_rules()
s.follow_UIC_rules()
f.facultyFunc()
s.studentFunc()
```

*  If we have not defined an **\__init__** method in a subclass, Python will call the constructor of the base class automatically.

* **In fact, Python always starts looking for methods in the actual subclass type first, and if it doesn’t find anything, it starts looking at the methods in the subclass’s base classes, one by one in the order they are specified in the tuple (here we only have 1 base class, but you can have multiple base classes) in the class definition.**

* Subclass can use those attributes (including fields and methods) in super class **as if they were defined in the subclass**, including super class's constructor. This can be illustrated below. You can see although `Faculty` class does not define `parentFunc()` method, the function still exists in the class.

```python
dir(f)
```

* It is fine to initialize subclass directly without using parent class's __init__ method, however side effect is it brings a lot of repetition. For example below, both `Faculty` and `Student` classes need to add `self.name=name` and `self.age = age` in their constructor. 

#### Example 5.2

```python
class SchoolMember: # base class
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def follow_UIC_rules(self):
        print ('%s should follow UIC rules 2021 version (updated)'% self.name)      


class Faculty(SchoolMember): # subclass, inherit from "SchoolMember" class
    def __init__(self, name, age, salary):
        # initialize subclass directly without using parent class's __init__ method
        self.name = name
        self.age = age
        self.salary = salary


class Student(SchoolMember):
    def __init__(self, name, age, marks):
        # initialize subclass directly without using parent class's __init__ method
        self.name = name
        self.age = age
        self.marks = marks
```

```python
f = Faculty('Eric', 30, 1000)
s = Student('Bob', 20, 90)
f.follow_UIC_rules()
s.follow_UIC_rules()
```

* A more elegant way to initialize subclass by calling superclass's constructor, like Example below:

#### Example 5.3

```python
class SchoolMember: # base class
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def follow_UIC_rules(self):
        print ('%s should follow UIC rules 2021 version (updated)'%self.name)       


class Faculty(SchoolMember): # subclass, inherit from "SchoolMember" class
    def __init__(self, name, age, salary):
        # initialize subclass by calling superclass's constructor
        SchoolMember.__init__(self, name, age)
        #super(Faculty, self).__init__(name, age)
        #super().__init__(name, age)
        self.salary = salary


class Student(SchoolMember):
    def __init__(self, name, age, marks):
        # initialize subclass by calling superclass's constructor
        SchoolMember.__init__(self, name, age) 
        #super(Student, self).__init__(name, age)
        #super().__init__(name, age)
        self.marks = marks

```

```python
f = Faculty('Eric', 30, 1000)
f.follow_UIC_rules()
```

We observe that the `__init__` method of the base class is explicitly called using the **self** variable so that we can initialize the base class part of an instance in the subclass. This is quite natural, because `name` and `age` belong to `SchoolMember` class, which is commonly existed in any subclass derived from `SchoolMember`.

### 5.2 Overriding

In Python method overriding occurs simply defining in the child class a method with the same name of a method in the parent class.

```python
class SchoolMember: 
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
        
class Faculty(SchoolMember): 
    def __init__(self, name, age, salary):
        SchoolMember.__init__(self, name, age)
        #super(Faculty, self).__init__(name, age)
        #super().__init__(name, age)
        self.salary = salary
    
    def get_evaluate(self):
        print ('To ensure the teaching and learning quality, every one needs to be evaluated.')
        print ('Teacher %s is evaluated by Students\' feedback'% self.name)

class Student(SchoolMember):
    def __init__(self, name, age, marks):
        SchoolMember.__init__(self, name, age) 
        #super(Student, self).__init__(name, age)
        #super().__init__(name, age)
        self.marks = marks
               
    def get_evaluate(self):
        print ('To ensure the teaching and learning quality, every one needs to be evaluated.')
        print ('Student %s is evaluated by Students\' final exam result'% self.name)
```

```python
f = Faculty('Eric', 30, 1000)
s = Student('Bob', 20, 80)
l = [f,s]
for obj in l:
    obj.get_evaluate()
```

* We can find that `Faculty` and `Student` class's `get_evaluate` function needs to perform a common task, to print out a sentence "To ensure the teaching and learning quality, every one needs to be evaluated". Actually, this is the  common task for each `SchoolMember`. A better way is to put the implementation of this common task (print the sentence) under `SchoolMember` class, as follows:


```python
class SchoolMember: 
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def get_evaluate(self): 
        print ('To ensure the teaching and learning quality, every one needs to be evaluated. (updated)')
        
class Faculty(SchoolMember): 
    def __init__(self, name, age, salary):
        SchoolMember.__init__(self, name, age)
        self.salary = salary
    
    def get_evaluate(self):
        SchoolMember.get_evaluate(self)
        print ('Teacher %s is evaluated by Students\' feedback'% self.name)

class Student(SchoolMember):
    def __init__(self, name, age, marks):
        SchoolMember.__init__(self, name, age) 
        self.marks = marks
               
    def get_evaluate(self):
        SchoolMember.get_evaluate(self)
        print ('Student %s is evaluated by Students\' feedback'% self.name)
```

```python
f = Faculty('Eric', 30, 1000)
s = Student('Bob', 20, 80)
l = [f,s]
for obj in l:
    obj.get_evaluate()
```

**Explanation**

* Remember that Python always starts looking for methods in the actual subclass type first, and if it doesn’t find anything, it starts looking at the methods in the subclass’s base classes. In both `Teacher` and `Student` class we define a `get_evaluate` method, when Teacher and Student objects call it, Python find that `get_evaluate` already exists under the `Teacher` and `Student` class, Python stops looking for `get_evaluate` method in `SchoolMember` class.

* However, we can find that **under** `Teacher` and `Student` class, `SchoolMember` class's `get_evaluate` method is called. This is because the method is jointly used in all `SchoolMember`'s subclasses, e.g. `Teacher` and `Student`.

* The major advantage of lifting the jointly used method up to base class is, you only need to update once of the method and the modification can take effect on all the subclasses.

* Why do we need to override method in parent class? One reason is you may want different subclasses have their own special implementation of the method to show their different functionality in your subclass.

## 6. Data Hiding

* In Python sometimes you will want to hide the object’s attributes outside the class definition. For that, you can use double underscore before the attributes name and those attributes will not be directly visible outside.

|  Name | Notation  | Behavior  |
|---|---|---|
|  name | Public  | Can be accessed from inside and outside  |
|  \_name | Protected  | Like a public member, but they shouldn't be directly accessed from outside.  |
| \__name  | Private  | Can't be seen and accessed from outside  |

#### Example 6.1

```python
class PC:
    __CPU = 0
    CPU = 0
    def access_CPU_by_usb_interface(self, increment):
        self.__CPU += increment
        print (self.__CPU)

lenovo = PC()
lenovo.access_CPU_by_usb_interface(2)
lenovo.access_CPU_by_usb_interface(5)
print (lenovo.CPU)
print (lenovo.__CPU)
```

* You see that `__hiddenVariable` is not accessible. Actually, Python protects the members by internally changing the names to include the class name.

```python
class PC:
    __CPU = 0
    CPU = 0
    def access_CPU_by_usb_interface(self, increment):
        self.__CPU += increment
        print (self.__CPU)

lenovo = PC()
lenovo.access_CPU_by_usb_interface(2)
lenovo.access_CPU_by_usb_interface(5)
print (lenovo.CPU)
print (lenovo._PC__CPU)
```
