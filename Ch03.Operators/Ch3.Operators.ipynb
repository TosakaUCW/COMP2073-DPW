| ---|---|---|
| + | Add two operands or unary plus | x + y +2 |

Check the following example and observe how they behave.

**Task 1**: In the code cell below, write the Python function `formula` which 3 arguments, `x`, `y`, and `z`, and returns the value of: $\lfloor\frac{x^{2y}+z}{\sqrt{z+1}}\rfloor$

````python
def formula(x, y, z):
    result = (x ** (2 * y) + z) / ((z + 1) ** 0.5)
    return int(result) if result >= 0 else int(result) - 1
````

````python
assert formula(2, 2, 3) == 9
assert formula(2, 2, 1) == 12
````

## Assignment Operators

They are almost the same as those in C except that, in Python, there is **NO** ``++`` or ``--``.

**Task 2**: Read the following C code:
```c
#include <stdio.h>
int nextAge(currentAge){
    if(currentAge<0 || currentAge>200){
        puts("Invalid age for a human being!");
        return 0;
    }
    return ++currentAge;
}
```

Convert it to Python in the code cell below:

````python
def nextAge(currentAge):
    if currentAge < 0 or currentAge > 200:
        print("Invalid age for a human being!")
        return 0
    return currentAge + 1
````

## Special Operators

### Identity Operators

``is`` and ``is not`` are the identity operators in Python. They are used to check if two values (or variables) are located on the same part of the memory. Two variables that are equal does not imply that they are identical.

The following example shows the use of these two operators.


````python
# Example of Identity Operators
x = 2
y = 2
print(x is y)

x = "Hello"
y = "Hello"
print(x is y)

x = [1, 2]
y = [1, 2]
print(x is y)
````

In this example, when ``x`` and ``y`` are both integers or strings, if they are equal in value, they are identical. However, when ``x`` and ``y`` are of complex data types such as *list* (the last case), the situation is different. Since lists are mutable (can be changed), interpreter locates them separately in memory. Therefore the two lists are equal but not identical.

The mutable data types will be introduced in later chapters.

### Membership operators

````python
print(0 in range(5))
print(5 in range(5))
print(1 in [1, 2, 3])
print([1, 2] in [1, 2, 3])
````
