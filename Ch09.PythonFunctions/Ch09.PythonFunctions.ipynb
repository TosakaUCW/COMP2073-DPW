## 1. Introduction

In Python, a function is a group of related statements that performs a specific task.

Functions help break our program into smaller and modular chunks. As our program grows larger and larger, functions make it more organized and manageable.

Furthermore, it avoids repetition and makes the code reusable.

We have introduced some basic knowledge about Python functions. In this chapter, let's start with an example:

````python
import math

def longest_side(a, b):
    """
    Function to find the length of the longest side of a right triangle.

    :arg a: Side a of the triangle
    :arg b: Side b of the triangle

    :return: Length of the longest side c as float
    """
    return math.sqrt(a*a + b*b)
````

**Task 1:** Complete the function `isValidDate` that accepts a string `date` and decide whether the date is valid or not. `date` takes a form `yyyy-mm-dd` and the year is greater than 0.  The function returns `True` if the date is valid and `False` otherwise.

For example, `isValidDate('2020-02-29')` returns `True` and `isValidDate('2021-02-29')` returns `False`.

* Note 1: You should strictly follow the example in the function `longest_side` and **write the docstring** of the function.
* Note 2: You are **not allowed** to import any modules.
* Note 3: You can assume that the format of the date is always correct.


````python
def isValidDate(date):
    """
    Determines whether a given date string in 'yyyy-mm-dd' format is valid.

    A valid date:
    - Has a year greater than 0.
    - The month is between 1 and 12.
    - The day is valid for the given month, considering leap years.

    Parameters:
    date (str): A string representing a date in the format 'yyyy-mm-dd'.

    Returns:
    bool: True if the date is valid, False otherwise.
    """
    # Split the date string into year, month, and day
    year, month, day = map(int, date.split('-'))
    
    # Check if the year is greater than 0
    if year <= 0:
        return False

    # Days in each month (index 0 is unused for direct mapping)
    days_in_month = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

    # Adjust for leap year
    if month == 2 and (year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)):
        days_in_month[2] = 29  # Leap year February has 29 days

    # Check if month and day are valid
    if 1 <= month <= 12 and 1 <= day <= days_in_month[month]:
        return True

    return False
````

````python
assert isValidDate('2020-02-29') == True
assert isValidDate('2024-03-31') == True
assert isValidDate('2024-04-31') == False
assert isValidDate('2024-13-01') == False
````

## 2. Function Arguments (Parameters)

* You can call a function by using the following types of formal arguments:
  * Required arguments
  * Keywords arguments
  
### 2.1 Required arguments

Required arguments are the arguments passed to a function in the correct positional order. Here, the number of arguments in the function call should match exactly the function definition. Below is an example.

````python
def printme(a, b):
    # Thih prints a passed string into this function.
    print('a is ', a, ' b is ', b)


# Now you can call `printme` function.
# Note, '1' and '2' are exactly matching to 'a' and 'b' here.
printme(1, 2)
````

### 2.2 Keyword Arguments

* Keyword arguments are related to the function calls. When you use keyword arguments in a function call, the caller identifies the arguments by the parameter name.

* This allows you to place arguments **out of order** because the Python interpreter is able to use the keywords provided to match the values with the parameters. 

````python
# print "a is:  9 , b is:  1"
printme(a=9, b=1)

# print "a is:  9 , b is:  1"
printme(b=1, a=9)
````

### 2.3 Default arguments

Function arguments can have default values in Python.
* If a parameter has a default value, it's **not mandatary** in the function call.
* Any number of arguments in a function can have a default value
* Once we have a default argument, *all the arguments to its right* must also have default values.

Observe the following example:

````python
def printme(a=9, b=1):
    print('a is: ', a, ', b is: ', b)


# Compare following function calls
# print "a is:  4 , b is:  5"
printme(4, 5)

# print "a is:  4 , b is:  1"
printme(4)

# print "a is:  9 , b is:  1"
printme()

# print "a is:  9 , b is:  5"
printme(b=5)
````

### 2.4 Variable-length arguments

* When calling the function, the number of the argument can vary.
* In the function definition, we use variable-length arguments.

Syntax for a function with non-keyword variable arguments is:

```python
def function_name([formal_args, ] *var_args_tuple):
    function_suite
    return [expression]
```

* An asterisk(\*) is placed before the variable name that holds the values of all non-keyword variable arguments. This tuple remains empty if no additional arguments are specified during the function call.

````python
def printinfo(a, *vartuple):
    # This prints a variable passed arguments.
    print('Output is: ')
    print(a)
    for var in vartuple:
        print(var)
    # print(*vartuple)


# Sometimes, the number of values following 'hi' is not fixed.
printinfo('hi', 1, 2, 3, 4)

another_name = printinfo
another_name('hi', 1, 2)

print(another_name == printinfo)
````

## 4. Anonymous Functions and the Lambda Operator

These functions are called anonymous because they are not declared in the standard manner by using the `def` keyword. You can use the `lambda` keyword to create small anonymous functions.
* Lambda forms can take any number of arguments but return just one value in the form of an expression. They cannot contain commands or multiple expressions.
* An anonymous function cannot be a direct call to print because lambda requires an expression
* Lambda functions have their own local namespace and cannot access variables other than those in their parameter list and those in the global namespace.
* A lambda function is a value so you can store it in a variable, just like any other value.
  
**Syntax**
```python
lambda [arg1 [, arg2, ..., argn]]: expression
```

**Example**

````python
# Function definition is here
sum_lambda1 = lambda arg1, arg2: arg1 + arg2

# Now you can call sum as a function
print ("Value of total : ", sum_lambda1( 10, 20 ))
print ("Value of total : ", sum_lambda1( 20, 20 ))
````

````python
# This gives the function another name
sum_lambda2 = sum_lambda1

# Now you can call the same function using sum_lambda2
print ("Value of total : ", sum_lambda2( 10, 20 ))
print ("Value of total : ", sum_lambda2( 20, 20 ))
````

````python
# The two names refer to the same function, they are the same!
sum_lambda2 is sum_lambda1
````

## 5. Pass by Reference

* All parameters (arguments) in Python language are passed by reference, which means if you change what a parameter refers to within a function, the change also reflects back in the calling function.

**Example** - The example below shows the function maintaining the reference of the passed object and replacing values in the same object.

````python
# Function definition is here.
def changed(mylist):
    # This change a passed list into this function
    mylist[:] = [1, 2, 3]
    print('Values inside the function: ', mylist)


# Now you can call the changed function.
mylist = [10, 20, 30]
changed(mylist)
print('Values outside the function: ', mylist)
````

However, if you just assign a new value to the argument, the object that the argument originally refers to is not changed. Below is another example.

````python
def nochange(mylist):
    # This does not change the list passed to the function
    mylist = [1, 2, 3]
    print('Values inside the function: ', mylist)
    
# Now you can call the nochange function.
mylist = [10, 20, 30]
nochange(mylist)
print('Values outside the function: ', mylist) 
````

## 6. Scope of Variables

The variables in a program may not be accessible from all the locations in that program. 

The **scope** of a variable depends on where you have declared it, and determines the part of the program where you can access it. There are two basic scopes of variables in Python.
* Local variables
  * A local variable is defined inside of a function.
  * Object outside the "scope" of the function will not be able to access the variable.
  
Below is an example of a local variable:

````python
def getPai():
    pai = 3.14
    return pai
    
getPai()
# pai is defined in getPai and cannot be accessed outside of getPai
# therefore, this statement causes a NameError
print(pai)
````

* Global variables
  * A global variable is defined outside of any function.
  * A global variable can be accessed inside of a function.
  * Use the  `global` keyword inside of a function before accessing a global variable.
  
 Below are some examples of global variables

````python
def increase(n):
    # must use the global keyword to indicate that it's a global variable
    # without it there will be an UnboundLocalError
    global total
    # Since 'total' is explicitly defined as globals,
    # change of the variable will affect the value of the global variable
    total += n
    print('Inside the function, total: ', total)

# total is defined outside of any function, it is a global variable
total = 1
# Now you can call the `increase` function.
print("Initial total: ", total)
increase(1)
print('Outside the function, total: ', total)
increase(1)
print('Outside the function, total: ', total)
````

**Task 2:** Coplete the function `isInTriangle` which accepts a point, `p`, in a 2 dimensional coordination system, and decides whether `p` is inside of a triangle or not. Note that `p` is represented using a tuple `(x, y)`. The triangle is defined by three points, and they are global tuples, `a`, `b` and `c`, respectively.

The function should return `True` if the point is on or inside of the triangle, and `False` otherwise.

````python
def isInTriangle(p):
    """
    Determines whether a point p is inside or on the boundary of a triangle
    defined by global points a, b, and c.

    Parameters:
    p (tuple): A tuple (x, y) representing the point to check.

    Returns:
    bool: True if the point is inside or on the triangle, False otherwise.
    """
    def area(x1, y1, x2, y2, x3, y3):
        """Helper function to compute the area of a triangle using determinant formula."""
        return abs(x1*(y2 - y3) + x2*(y3 - y1) + x3*(y1 - y2)) / 2.0
    
    global a, b, c

    A = area(*a, *b, *c)
    A1 = area(*p, *b, *c)
    A2 = area(*a, *p, *c)
    A3 = area(*a, *b, *p)

    return A == (A1 + A2 + A3)
````

````python
a = (0, 0)
b = (3, 3)
c = (4, 0)
assert isInTriangle((0,0)) == True
assert isInTriangle((3,0)) == True
assert isInTriangle((3,2)) == True
assert isInTriangle((3,4)) == False
assert isInTriangle((3,-2)) == False
````

**Task 3:** Complete the function `decimal_2_binary` that accepts a positive integer, `number`, and returns a string, which contains the binary representation of `number`.

*Note: You are **not allowed** to use any library functions such as `bin`.*

````python
def decimal_2_binary(number):
    binary = ""
    while number > 0:
        binary = str(number % 2) + binary
        number //= 2
    return binary if binary else "0"
````

````python
assert decimal_2_binary(14) == '1110'
assert decimal_2_binary(1) == '1'
````

**Task 4:** Complete the function `find_rep_str` which accepts a string, `s`, and finds all the repeated characters which are sub-strings of `s`. The substrings should be added to a global set, `result`.

Hint: Using `dir(str)` to find the built-in function of Python object `str`,and find the way to solve the problem.

````python
def find_rep_str(s):
    global result
    
    for i in range(len(s)):
        for j in range(i + 1, len(s)):
            if s[i] != s[j]:
                break
            result.add(s[i:j+1])
````

````python
result = set()
find_rep_str('http://www.google.com')
assert result == {'tt', '//', 'www', 'ww', 'oo'}
find_rep_str('byebye')
assert result == {'tt', '//', 'www', 'ww', 'oo'}
find_rep_str('oh  oh')
assert result == {'tt', '//', 'www', 'ww', 'oo', '  '}
````

**Task 5:** Complete the function `cmpt_ticket` that computes the ticket price of a park based on the `original_price`. The policy of ticketing is:
  * The ticket price on weekends is 1.2 times of that on weekdays. By default, it is assumed that the day is a weekday.
  * The ticket price for a child is half of that for an adult. By default, it is assumed that the customer is an adult.
  
*Note: You have to modify the function header to set the default value of parameters.*

````python
def cmpt_ticket(original_price, is_weekend=False, is_child=False):
    price = original_price * 1.2 if is_weekend else original_price
    return price * 0.5 if is_child else price
````

````python
assert cmpt_ticket(100) == 100
assert cmpt_ticket(100, is_child=True) == 50
assert cmpt_ticket(100, is_weekend=True) == 120
assert cmpt_ticket(100, True, True) == 60
````
